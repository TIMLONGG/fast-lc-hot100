```python
# 42. 接雨水
# H
# 2
class Solution:
    def trap(self, height: List[int]) -> int:
        # 双指针
        n = len(height)
        l, r = 0, n - 1 
        pre = 0
        suf = 0
        ans = 0
        while l < r:
            pre = max(pre, height[l])
            suf = max(suf, height[r])
            if pre < suf:
                ans += pre - height[l]
                l += 1
            else:
                ans += suf - height[r]
                r -= 1

        return ans

# 依然没有完全理解l和r的含义。还需要巩固。
```

```python
# 239. 滑动窗口最大值
# H
# 3

class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        q = deque()
        ans = []
        for i, x in enumerate(nums):
            while q and x >= nums[q[-1]]:
                q.pop()

            q.append(i)

            left = i - k + 1
            if q[0] < left:
                q.popleft()
            
            if left >= 0:
                ans.append(nums[q[0]])

        return ans

```

```python
# 128. 最长连续序列
# M
# 4

class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:
        # 找开头
        st = set(nums)
        ans = 0
        for i in st:
            if i - 1 in st:
                continue
            
            y = i + 1
            while y in st:
                y += 1
                
            ans = max(ans, y - i)
        return ans

```

```python
# 76. 最小覆盖子串
# H
# 1

# 解法1：暴力
class Solution:
    def minWindow(self, s: str, t: str) -> str:
        cnt_s = Counter()
        cnt_t = Counter(t)

        ans_left, ans_right = -1, len(s)
        left = 0

        for right, c in enumerate(s):
            cnt_s[c] += 1
            while cnt_s >= cnt_t:
                if right - left < ans_right - ans_left:
                    ans_left, ans_right = left, right
                cnt_s[s[left]] -= 1
                left += 1
        
        return "" if ans_left < 0 else s[ans_left: ans_right + 1]

# 这里学到了“涵盖”关系可以直接使用两个Counter对比（大于小于）来判断。


# 解法2：单个变量记录满足条件的字母种类
class Solution:
    def minWindow(self, s: str, t: str) -> str:
        diff = defaultdict(int)  # diff表示cnt_s - cnt_t
        for c in t:
            diff[c] -= 1
        
        kinds = len(diff)  # t中字母的种类数量

        ans_left, ans_right = -1, len(s)
        ge_cnt = 0  # 满足要求的字母种类数量
        left = 0

        for right, c in enumerate(s):
            diff[c] += 1
            if diff[c] == 0:  # 刚满足条件
                ge_cnt += 1
            
            while ge_cnt == kinds:
                if right - left < ans_right - ans_left:
                    ans_left, ans_right = left, right  # 更新答案
            
                x = s[left]
                if diff[x] == 0:  # 由于left移动，恰好要不满足条件了
                    ge_cnt -= 1
                diff[x] -= 1
                left += 1
        
        return "" if ans_left < 0 else s[ans_left: ans_right + 1]
```

```python
# 53. 最大子数组和
# M
# 4

# 前缀和 需要复习
class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        
        ans = -inf
        min_pre_sum = pre_sum = 0
        for x in nums:
            pre_sum += x
            ans = max(ans, pre_sum - min_pre_sum)
            min_pre_sum = min(min_pre_sum, pre_sum)
        
        return ans

# 动态规划
class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        f = 0
        ans = -inf
        for i in nums:
            f = max(f + i, i)
            ans = max(ans, f)
        return ans

# 子序列相关问题，可以多想想能不能用前缀和。
```

```python
# 56. 合并区间
# M
# 4

class Solution:
    def merge(self, intervals: List[List[int]]) -> List[List[int]]:
        ans = []
        intervals.sort(key=lambda p: p[0])

        for x in intervals:
            l, r = x[0], x[1]
            if ans and ans[-1][1] >= l:
                ans[-1][1] = max(ans[-1][1], r)
            else:
                ans.append(x)

        return ans

# 核心思路，就是将每一个区间，按照左端点进行排序。
```
