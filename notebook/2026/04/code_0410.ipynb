```python
# 41. 缺失的第一个正数
# H
# 3 

class Solution:
    def firstMissingPositive(self, nums: List[int]) -> int:
        # 2026-04-10
        # Time 16:56
        n = len(nums)

        for i in range(n):
            while 1 <= nums[i] <= n and nums[nums[i] - 1] != nums[i]:
                j = nums[i] - 1
                nums[j], nums[i] = nums[i], nums[j]

        for i in range(n):
            if i != nums[i]-1:
                return i + 1
        return n + 1

# 遇到的问题是交换后的元素，无法被遍历到了
# 核心：while那边写成if了。导致把交换后的元素，无法被安排到对应的位置。
```

```python
# 560. 和为 K 的子数组
# M
# 4

class Solution:
    def subarraySum(self, nums: List[int], k: int) -> int:
        # 2026-04-10
        # Time 07:08
        pre = 0
        ans = 0     
        dic = defaultdict(int)
        n = len(nums)
        dic[pre] = 1 
        for i in range(n):
            pre = pre + nums[i]
            ans += dic[pre-k]
            dic[pre] += 1
        return ans
        
```

```python
# 239. 滑动窗口最大值
# H
# 4

class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        # Time 06:25
        q = deque()
        ans = []
        for r, x in enumerate(nums):
            while q and x >= nums[q[-1]]:
                q.pop()
            
            q.append(r)

            l = r - k + 1
            if q[0] < l:
                q.popleft()

            if l >= 0:
                ans.append(nums[q[0]])
        
        return ans

# return 位置写错了
```
