```python
# 76. 最小覆盖子串
# H
# 5

class Solution:
    def minWindow(self, s: str, t: str) -> str:
        # 2026-04-13
        # Time 3:30

        diff = defaultdict(int)
        for i in t:
            diff[i] -= 1
        kinds = len(set(t))
        cnt = l = 0
        ans_l, ans_r = -1, len(s)
        for r, c in enumerate(s):
            diff[c] += 1
            if diff[c] == 0:
                cnt += 1
            
            while cnt == kinds:
                if r - l < ans_r - ans_l:
                    ans_l, ans_r = l, r
                diff[s[l]] -= 1
                if diff[s[l]] == -1:
                    cnt -= 1
                l += 1
        return "" if ans_l == -1 else s[ans_l:ans_r+1]
```

```python
# 42. 接雨水
# H
# 5

class Solution:
    def trap(self, height: List[int]) -> int:
        # 2026-04-13
        # Time 2:15

        pre = suf = l = ans = 0
        r = len(height) - 1

        while l < r:
            pre = max(pre, height[l])
            suf = max(suf, height[r])

            if pre < suf:
                ans += pre - height[l]
                l += 1
            
            else:
                ans += suf - height[r]
                r -= 1
        
        return ans
```

```python
# 3. 无重复字符的最长子串
# M
# 5

class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        # 2026-04-13
        # Time 3:51

        l = ans = 0
        st = set()
        for r, x in enumerate(s):
            while x in st:
                st.remove(s[l])
                l += 1

            st.add(x)

            ans = max(ans, r - l + 1)
        return ans
```

```python
# 226. 翻转二叉树
# E
# 5

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def invertTree(self, root: Optional[TreeNode]) -> Optional[TreeNode]:
        # 2026-04-13
        # Time 5:26
        if not root:
            return None
        root.left, root.right = root.right, root.left
        self.invertTree(root.left)
        self.invertTree(root.right)
        return root
```

```python
# 543. 二叉树的直径
# E
# 3

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def diameterOfBinaryTree(self, root: Optional[TreeNode]) -> int:
        # 2026-04-13
        # Time 17:59
        ans = 0
        def dfs(node):
            if not node:
                return -1
            nonlocal ans
            l_depth = dfs(node.left) + 1
            r_depth = dfs(node.right) + 1
            ans = max(ans, l_depth + r_depth)
            return max(l_depth, r_depth) # 当前子树最大链长
        dfs(root)
        return ans
    
# return -1 表示链长，如果return 为0，则表示链长+1了。
```

```python
# 102. 二叉树的层序遍历
# M
# 4

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def levelOrder(self, root: Optional[TreeNode]) -> List[List[int]]:
        # 2026-04-13
        # Time 11:34

        # 一个队列解决
        if root is None:
            return []
        
        ans = []
        q = deque([root])
        while q:
            vals = []
            for _ in range(len(q)):
                node = q.popleft()
                vals.append(node.val)
                # 如果是None就不加入队列，这样就避免后续出现空的vals加在ans最后
                if node.left:
                    q.append(node.left)
                if node.right:
                    q.append(node.right)
            ans.append(vals)
        return ans

        
```

```python
# 108. 将有序数组转换为二叉搜索树
# E
# 3

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def sortedArrayToBST(self, nums: List[int]) -> Optional[TreeNode]:
        # 2026-04-13
        # Time 21:48
        
        if not nums:
            return None
        m = len(nums) // 2
        left = self.sortedArrayToBST(nums[:m])
        right = self.sortedArrayToBST(nums[m+1:])
        return TreeNode(nums[m], left, right)

# 核心思路就是不断寻找中点作为子树的根节点。
```

```python
# 70. 爬楼梯
# E
# 4 4

class Solution:
    def climbStairs(self, n: int) -> int:
        # 2026-04-13
        # Time 6:51
        # 空间复杂度 O(n)
        dp = [0] * (n + 1)
        dp[0] = 1
        dp[1] = 1
        for i in range(2, n+1):
            dp[i] = dp[i-1] + dp[i-2]
        return dp[-1]


class Solution:
    def climbStairs(self, n: int) -> int:
        # 2026-04-13
        # Time 2:56
        # 空间复杂度 O(1)
        f0 = f1 = 1
        for _ in range(2, n + 1):
            new_f = f1 + f0
            f0 = f1
            f1 = new_f
        return f1

# 第二种做法要好好掌握。
```
