```python
# 98. 验证二叉搜索树
# M
# 4
# 上次备注: 中序遍历解法薄弱

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    pre = -inf
    def isValidBST(self, root: Optional[TreeNode]) -> bool:
        # 2026-04-16
        # Time: 03:54

        # 中序
        if root is None:
            return True

        if not self.isValidBST(root.left):
            return False

        if root.val <= self.pre:
            return False
        
        self.pre = root.val
        return self.isValidBST(root.right)
```

```python
# 25. K 个一组翻转链表
# H
# 4
# 上次备注: 掌握 pre 的迭代更新逻辑

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reverseKGroup(self, head: Optional[ListNode], k: int) -> Optional[ListNode]:
        # 2026-04-16
        # Time 10:49
        
        cnt = 0
        cur = head
        while cur:
            cur = cur.next
            cnt += 1
        
        pre = None
        cur = head
        p0 = dummy = ListNode(next=head)
        while cnt >= k:
            cnt -= k

            for _ in range(k):
                nxt = cur.next
                cur.next = pre
                pre = cur
                cur = nxt

            nxt = p0.next
            p0.next.next = cur
            p0.next = pre
            p0 = nxt

        return dummy.next

# p和dummy都是指针，指向指向head（指向的那个对象）的对象
```

```python
# 189. 轮转数组
# M
# 3
# 上次备注: 翻转数组解法忘记取模

class Solution:
    def rotate(self, nums: List[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        # 2026-04-16
        # Time 2:58
        # 记得取模！！！
        
        def reverse(l, r):
            while l < r:
                nums[l], nums[r] = nums[r], nums[l]
                l += 1
                r -= 1
        n = len(nums)
        k %= n
        reverse(0, n - 1)
        reverse(0, k - 1)
        reverse(k, n - 1)

# 又忘了取模！！！！
# 另外，为什么nums可以直接用：
# Python 内部函数中，一旦出现赋值语句，该变量就会被判定为局部变量，而 nums[i] = ... 这种原地修改只是对已有对象内容的成员操作，会直接沿用外层变量的引用。
```

```python
# 92. 反转链表 II
# M
# 4
# 上次备注: 注意 p0.next.next 和 p0.next 的重指向

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reverseBetween(self, head: Optional[ListNode], left: int, right: int) -> Optional[ListNode]:
        # 2026-04-16
        # Time 
        p0 = dummy = ListNode(next=head)
        
        for _ in range(left - 1):
            p0 = p0.next
        
        head = p0.next
        pre = None
        for _ in range(right - left + 1):
            nxt = head.next
            head.next = pre
            pre = head
            head = nxt
        
        p0.next.next = head
        p0.next = pre

        return dummy.next
```

```python
# 230. 二叉搜索树中第 K 小的元素
# M
# 5

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def kthSmallest(self, root: Optional[TreeNode], k: int) -> int:
        # 2026-04-16
        # Time 7:27
        ans = 0
        def dfs(node: Optional[TreeNode]) -> None:
            nonlocal k, ans
            if node is None or k <= 0:
                return
            dfs(node.left)  # 左
            k -= 1
            if k == 0:
                ans = node.val  # 根
            dfs(node.right)  # 右
        dfs(root)
        return ans
```
