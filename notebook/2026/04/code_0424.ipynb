```python
# 198. 打家劫舍
# M
# 5
# 上次备注: O(1) 空间状态转移

class Solution:
    def rob(self, nums: List[int]) -> int:
        # 2026-04-24
        # Time 02:09

        f0, f1 = 0, 0
        # dp[i] = max(dp[i-1], dp[i-2] + nums[i])
        for x in nums:
            dp = max(f1, f0 + x)
            f0 = f1
            f1 = dp
        return f1
```

```python
# 49. 字母异位词分组
# M
# 4

class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        # 2026-04-24
        # Time 04:32

        cnt = defaultdict(list)
        for x in strs:
            i = "".join(sorted(x))
            cnt[i].append(x)
        return list(cnt.values())


b = sorted("bca")
type(b)  # list
# sorted返回对象是列表，需要转成str才能作key
```

```python
# 98. 验证二叉搜索树
# M
# 4

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    pre = -inf
    def isValidBST(self, root: Optional[TreeNode]) -> bool:
        # 2026-04-24
        # Time: 03:54

        if root is None:
            return True
        
        if not self.isValidBST(root.left):
            return False
        
        if root.val <= self.pre:
            return False
        self.pre = root.val
        return self.isValidBST(root.right)

# pre是类属性 要么用self.pre访问 要么Solution.pre去访问
# 更新的时候，要么继续更新类属性，要么绑定到实例属性self.pre上
```

```python
# 78. 子集
# M
# 5 4

class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:
        # 2026-04-24
        # Time 02:18

        path = []
        n = len(nums)
        ans = []
        # 选或不选
        def dfs(i):
            if i == n:
                ans.append(path.copy())
                return
            
            # 不选
            dfs(i + 1)
            # 选
            path.append(nums[i])
            dfs(i + 1)
            path.pop()

        dfs(0)
        return ans


class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:
        # 2026-04-24
        # Time 03:53

        # 枚举答案

        ans = []
        path = []
        n = len(nums)

        def dfs(i):
            ans.append(path.copy())
            for j in range(i, n):
                path.append(nums[j])
                dfs(j + 1)  # 选 nums[j] 意味着 i 到 j-1 都跳过不选，下一个数从 j+1 开始
                path.pop()
        
        dfs(0)
        return ans
```
