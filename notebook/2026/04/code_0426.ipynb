```python
# 300. 最长递增子序列
# M
# 5

class Solution:
    def lengthOfLIS(self, nums: List[int]) -> int:
        # 2026-04-26
        # Time 02:15
        
        g = []
        for i, x in enumerate(nums):
            j = bisect_left(g, x)
            if j == len(g):
                g.append(x)
            else:
                g[j] = x
        return len(g)
```

```python
# 41. 缺失的第一个正数
# H
# 5
# 上次备注: 易错点：重复数字判断遗漏，且交换后判断会越界

class Solution:
    def firstMissingPositive(self, nums: List[int]) -> int:
        # 2026-04-26
        # Time 6:13
        n = len(nums)
        for i in range(n):
            while 1 <= nums[i] <=n and nums[nums[i] - 1] != nums[i]:
                j = nums[i] - 1
                nums[i], nums[j] = nums[j], nums[i]
        
        for i in range(n):
            if i + 1 != nums[i]:
                return i + 1
        return n + 1
```

```python
# 2. 两数相加
# M
# 5 4
# 上次备注: 递归迭代都需要着重加强

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode], c:int=0) -> Optional[ListNode]:
        # 2026-04-26
        # Time 3:04

        # 迭代
        dummy = cur = ListNode()
        
        while l1 or l2 or c:
            s = c
            if l1:
                s += l1.val
                l1 = l1.next
            if l2:
                s += l2.val
                l2 = l2.next
            c = s // 10
            cur.next = ListNode(s % 10)
            cur = cur.next

        return dummy.next
            

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode], c:int=0) -> Optional[ListNode]:
        # 2026-04-26
        # Time 5:28
        # 递归
        if not l1 and not l2:
            return ListNode(c) if c else None
        
        if not l1:
            l1, l2 = l2, l1
        
        s = l1.val + c + (l2.val if l2 else 0)
        l1.val = s % 10

        l1.next = self.addTwoNumbers(l1.next, l2.next if l2 else None, s // 10)
        return l1
```
