```python
# 51. N 皇后
# H
# 5
# 上次备注: 索引点：Python 可用负数索引，其他语言不行

class Solution:
    def solveNQueens(self, n: int) -> List[List[str]]:
        # 2026-04-27
        # Time 06:32
        m = 2 * n - 1
        diag1 = [False] * m
        diag2 = [False] * m
        col = [0] * n
        visit = [False] * n
        ans = []
        def dfs(i):
            if i == n:
                ans.append(['.' * c + 'Q' + '.' * (n-1-c) for c in col])
                return
            
            for j in range(n):
                if not visit[j] and not diag1[i + j] and not diag2[i - j]:
                    visit[j] = diag1[i + j] = diag2[i - j] = True
                    col[i] = j
                    dfs(i + 1)
                    visit[j] = diag1[i + j] = diag2[i - j] = False
        dfs(0)
        return ans
```

```python
# 25. K 个一组翻转链表
# H
# 5

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reverseKGroup(self, head: Optional[ListNode], k: int) -> Optional[ListNode]:
        # 2026-04-27
        # Time 4:17
        
        dummy = p0 = ListNode(next=head)
        cnt = 0
        cur = head
        while cur:
            cur = cur.next
            cnt += 1
        pre = None
        cur = head
        while cnt >= k:
            cnt -= k
            for _ in range(k):
                nxt = cur.next
                cur.next = pre
                pre = cur
                cur = nxt
            
            nxt = p0.next
            p0.next.next = cur
            p0.next = pre
            p0 = nxt
        return dummy.next
```

```python
# 102. 二叉树的层序遍历
# M
# 4


# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def levelOrder(self, root: Optional[TreeNode]) -> List[List[int]]:
        # 2026-04-27
        # Time 8:26
        if not root:
            return []
        q = deque([root])
        ans = []
        while q:
            n = len(q)
            s = []
            for _ in range(n):
                node = q.popleft()
                if node.left:
                    q.append(node.left)
                if node.right:
                    q.append(node.right)
                s.append(node.val)
            ans.append(s)
        return ans
```
