```python
from typing import List
# 1. 两数之和
# E
# P
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        dic = dict()
        for i, x in enumerate(nums):
            if target-x in dic:
                return [dic[target-x], i]
            dic[x] = i
# 哈希即可
```

```python
# 49. 字母异位词分组
# E
# P
class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        dic = defaultdict(list)
        for i, x in enumerate(strs):
            dic["".join(sorted(x))].append(x)
        return list(dic.values())

# 这题主要卡在了数据格式，dic.values()类型是dict_values
# 还有：dic直接存原值即可，没必要存索引！！！
```

```python
# 128. 最长连续序列
# M
# R
class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:
        st = set(nums)  # 这一步非常关键 转换成哈希集合
        m = len(st)
        ans = 0
        for x in st:
            if x - 1 in st:
                continue
            y = x + 1
            while y in st:
                y += 1
            ans = max(ans, y - x)
            if ans * 2 >= m:  # ans 不可能变得更大 乘法比除法好在m为偶数且ans=m//2时可以及时退出
                break
        return ans

# 不要一开始就陷入细节
# 这题本来想建立一个hash表去按照前面两题的思路做，但是一开始就陷入细节，考虑当前数是开头还是结尾，还是中间的情况。
# 核心思路就是就把当前数当做开头，先找到每个独立顺序序列的开头，然后分别计算他们的长度，长度大于一半直接返回，否则统计最大长度作为答案。
```

```python
# 283. 移动零
# E
# P
class Solution:
    def moveZeroes(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        l = 0
        for r in range(len(nums)):
            if nums[r]:
                nums[r], nums[l] = nums[l], nums[r]
                l += 1

# 因为保持相对顺序，所以双指针是都在左侧初始化的，而不是相向移动。
```

```python
# 11. 盛最多水的容器
# M
# P
class Solution:
    def maxArea(self, height: List[int]) -> int:
        l, r = 0, len(height) - 1
        area = 0
        while l < r:
            area = max(area, min(height[l], height[r]) * (r - l))
            if height[l] < height[r]:
                l += 1
            else:
                r -= 1 
        return area

# 一次解决。核心：移动短边，才有可能找到容积更大的情况。
```

```python
# 15. 三数之和
# M
# R
class Solution:
    def threeSum(self, nums: list[int]) -> list[list[int]]:
        nums.sort()
        ans = []
        m = len(nums)

        for i in range(m - 2):
            x = nums[i]
            if i > 0 and x == nums[i - 1]:
                continue
            if x + nums[i+1] + nums[i+2] > 0:  # x只会更大，sum大于0只能break
                break
            if x + nums[-1] + nums[-1] < 0:  # x只会更大，sum小于0还有机会
                continue
            
            j = i + 1
            k = m - 1
            while j < k:
                s = x + nums[j] + nums[k]
                if s > 0:
                    k -= 1
                elif s < 0:
                    j += 1
                else:
                    ans.append([x, nums[j], nums[k]])
                    j += 1
                    while j < k and nums[j] == nums[j-1]:
                        j += 1
                    k -= 1
                    while j < k and nums[k] == nums[k+1]:
                        k -= 1
        return ans
            
# 主要卡在去重方面。以及我做了没有意义的倒序遍历，正序即可。两个优化我放在了最外面，其实针对每一个x，都需要进行这两个优化。
```

```python
# 42. 接雨水
# H
# P(遍历两次) R(遍历一次)
class Solution:
    def trap(self, height: List[int]) -> int:
        # 双指针 遍历一次
        n = len(height)
        l, r = 0, n - 1
        pre = 0
        suf = 0
        ans = 0
        while l < r:
            pre = max(pre, height[l])
            suf = max(suf, height[r])
            if pre < suf:
                ans += pre - height[l]
                l += 1
            else:
                ans += suf - height[r]
                r -= 1
        return ans

class Solution:
    def trap(self, height: List[int]) -> int:
        # 前后缀分解
        # 时空复杂度 O(n)
        n = len(height)
        pre_max = [0] * n
        pre_max[0] = height[0]
        for i in range(1, n):
            pre_max[i] = max(pre_max[i - 1], height[i])
        
        suf_max = [0] * n
        suf_max[-1] = height[-1]
        for i in range(n - 2, -1, -1):
            suf_max[i] = max(suf_max[i + 1], height[i])
        
        ans = 0
        for h, pre, suf in zip(height, pre_max, suf_max):
            ans += min(pre, suf) - h
        
        return ans

# 我的解法依然是第一次做这道题时想到的遍历两次，但是总归得掌握最优解法。
# 前后缀分解是前置算法，虽然需要遍历三次，但掌握之后可以更好地理解双指针遍历一次的解法。
```

```python
# 3. 无重复字符的最长子串
# M
# R
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        l = 0
        ans = 0
        st = set()
        for r, c in enumerate(s):
            while c in st:
                st.remove(s[l])
                l += 1
            st.add(c)
            ans = max(ans, r - l + 1)
        return ans
        
# 子字符串 是字符串中连续的 非空 字符序列
# 有集合和字典两种写法，集合写法我觉得更简洁、直观一些。
# 这题虽然写出来了，但花了不少时间，还需要加强。
```

```python
# 438. 找到字符串中所有字母异位词
# M
# P(先用的暴力，后面改成了定长窗口)

# 定长滑窗
class Solution:
    def findAnagrams(self, s: str, p: str) -> List[int]:
        n, m = len(s), len(p)
        ans = []
        l = 0
        cntp = Counter(p)
        cnts = Counter(s[:m-1])
        for r in range(m - 1, n):
            cnts[s[r]] += 1
            if cnts == cntp:
                ans.append(l)
            cnts[s[l]] -= 1
            l += 1
        return ans

# 不定长滑窗 需要复习
class Solution:
    def findAnagrams(self, s: str, p: str) -> List[int]:
        cnt = Counter(p)
        ans = []

        left = 0
        for right, c in enumerate(s):
            cnt[c] -= 1
            while cnt[c] < 0:
                cnt[s[left]] += 1
                left += 1
            if right - left + 1 == len(p):
                ans.append(left)
        
        return ans


# 后面需要复习定长滑窗的做法。
# 定长滑窗模板就是先取长度k-1（上面的[:m-1]），这样方便循环（入-更新-出）
# 下面是灵神的模板，跟我的方法稍有不同：
class Solution:
    def maxVowels(self, s: str, k: int) -> int:
        ans = vowel = 0
        for i, c in enumerate(s):  # 枚举窗口右端点 i
            # 1. 右端点进入窗口
            if c in "aeiou":
                vowel += 1

            left = i - k + 1  # 窗口左端点
            if left < 0:  # 窗口大小不足 k，尚未形成第一个窗口
                continue

            # 2. 更新答案
            ans = max(ans, vowel)

            # 3. 左端点离开窗口，为下一个循环做准备
            if s[left] in "aeiou":
                vowel -= 1
        return ans
```

# Subarray

```python
# 560. 和为 K 的子数组
# M
# P(R)
# 耗时 8mins
class Solution:
    def subarraySum(self, nums: List[int], k: int) -> int:
        n = len(nums)
        pre_sum = [0] * (n + 1)
        for i in range(n):
            pre_sum[i + 1] = pre_sum[i] + nums[i]
        ans = 0
        dic = defaultdict(int)
        for i in pre_sum:
            ans += dic[i - k]
            dic[i] += 1
        return ans

# 这题只要想到要用前缀和，就会自然而然想到哈希，然后就简单了。偶尔复习一下。
```

```python
# 239. 滑动窗口最大值
# H
# RR 一点做不出来 需要重点理解
class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        q = deque()
        ans = []
        for i, x in enumerate(nums):
            while q and x >= nums[q[-1]]:
                q.pop()
            
            q.append(i)

            if i - q[0] >= k:
                q.popleft()
            
            if i >= k -1:
                ans.append(nums[q[0]])
        return ans
```
