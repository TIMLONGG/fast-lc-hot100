```python
# 92. 反转链表 II
# M
# 5


# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reverseBetween(self, head: Optional[ListNode], left: int, right: int) -> Optional[ListNode]:
        # 2026-05-01
        # Time 03:45

        dummy = p0 = ListNode(next=head)

        for _ in range(left - 1):
            p0 = p0.next
        
        cur = p0.next
        pre = None
        for _ in range(right - left + 1):
            nxt = cur.next
            cur.next = pre
            pre = cur
            cur = nxt
        
        p0.next.next = cur
        p0.next = pre

        return dummy.next
```

```python
# 102. 二叉树的层序遍历
# M
# 4

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def levelOrder(self, root: Optional[TreeNode]) -> List[List[int]]:
        # 2026-05-02
        # Time 5:12

        ans = []
        if not root:
            return []
        q = deque([root])
        while q:
            n = len(q)
            cur = []
            for i in range(n):
                node = q.popleft()
                if node.left:
                    q.append(node.left)
                if node.right:
                    q.append(node.right)
                cur.append(node.val)
            ans.append(cur)
        return ans
        
# 需要注意空树的判定
```

```python
# 142. 环形链表 II
# M
# 4
# 上次备注: Floyd判圈：相遇点到入口距离=起点到入口距离

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, x):
#         self.val = x
#         self.next = None

class Solution:
    def detectCycle(self, head: Optional[ListNode]) -> Optional[ListNode]:
        # 2024-05-02
        # Time 4:31

        fast = slow = head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            if slow == fast:
                slow = head
                while True:
                    if slow == fast:
                        return slow 
                    slow = slow.next
                    fast = fast.next
       
        return None

# 如果相遇点就在head节点，直接返回，所以要把if放在前面。
```

```python
# 74. 搜索二维矩阵
# M
# 3

class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:
        # 2026-05-02
        # Time

        m, n = len(matrix), len(matrix[0])
        left, right = 0, m * n - 1
        while left <= right:
            mid = (left + right) // 2
            x = matrix[mid // n][mid % n]
            if x == target:
                return True
            if x < target:
                left = mid + 1
            else:
                right = mid - 1
        return False

# 关键在坐标转换，也可以使用之前的排除法来做，不过二分复杂度一般更低一些
```
