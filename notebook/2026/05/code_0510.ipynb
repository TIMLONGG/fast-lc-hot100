```python
# 72. 编辑距离
# M
# 3

class Solution:
    def minDistance(self, word1: str, word2: str) -> int:
        # 2026-05-10
        # Time 3:20

        @cache
        def dfs(i, j):
            if i < 0:
                return j + 1
            if j < 0:
                return i + 1
            
            if word1[i] == word2[j]:
                return dfs(i-1, j-1)
            
            else:
                return min(dfs(i-1, j), dfs(i, j-1), dfs(i-1, j-1)) + 1
            
        
        return dfs(len(word1)-1, len(word2)-1)
```
