```python
# 41. 缺失的第一个正数
# H
# 5

class Solution:
    def firstMissingPositive(self, nums: List[int]) -> int:
        # 2026-04-11
        # Time 07:38

        n = len(nums)    
        for i in range(n):
            while 1 <= nums[i] <= n and nums[nums[i] - 1] != nums[i]:
                j = nums[i] - 1
                nums[i], nums[j] = nums[j], nums[i]
        
        for i in range(n):
            if nums[i] != i + 1:
                return i + 1
        return n + 1
```

```python
# 128. 最长连续序列
# M
# 5

class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:
        # 2026-04-11
        # Time 06:41
        st = set(nums)
        ans = 0
        for x in st:
            if x - 1 in st:
                continue

            y = x + 1
            while y in st:
                y += 1
            ans = max(ans, y - x)
        return ans
```

```python
# 160. 相交链表
# E
# 4 

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, x):
#         self.val = x
#         self.next = None

class Solution:
    def getIntersectionNode(self, headA: ListNode, headB: ListNode) -> Optional[ListNode]:
        # 2026-04-11
        # Time 08:03

        la = lb = 1
        cur = headA
        while cur.next:
            cur = cur.next
            la += 1
        cur = headB
        while cur.next:
            cur = cur.next
            lb += 1

        if la >= lb:
            for i in range(la - lb):
                headA = headA.next
        else:
            for i in range(lb - la):
                headB = headB.next

        for i in range(min(la, lb)):
            if headA is headB:
                return headA
            headA = headA.next
            headB = headB.next
            
        
        return None

# 官方做法是让先走完的去从头走另外一个链表，其实原理都是一致的，只不过实现起来更加优雅。
class Solution:
    def getIntersectionNode(self, headA: ListNode, headB: ListNode) -> Optional[ListNode]:
        p, q = headA, headB
        while p is not q:
            p = p.next if p else headB
            q = q.next if q else headA
        return p
```

```python
# 206. 反转链表
# E
# 3

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        # 2026-04-11
        # Time 14:36
        pre = None
        
        while head:
            nxt = head.next
            head.next = pre  # 头插法
            pre = head
            head = nxt
        
        return pre

```
