```python
# 76. 最小覆盖子串
# H
# 4

class Solution:
    def minWindow(self, s: str, t: str) -> str:
        # 2026-04-12
        # Time 8:51
        
        diff = defaultdict(int)
        for c in t:
            diff[c] -= 1
        kinds = len(set(t))
        cnt = 0
        l = 0
        ans_left, ans_right = -1, len(s)
        for r, c in enumerate(s):
            diff[c] += 1
            if diff[c] == 0:
                cnt += 1
            
            while cnt == kinds:
                if r - l < ans_right - ans_left:
                    ans_left, ans_right = l, r
                diff[s[l]] -= 1
                if diff[s[l]] < 0:
                    cnt -= 1
                l += 1
        
        return "" if ans_left == -1 else s[ans_left:ans_right+1]

# s[l]写成c了，然后diff[s[l]]-=1写成+=1了，还是不太熟练。不过逻辑比较清晰。
```

```python
# 42. 接雨水
# H
# 5

class Solution:
    def trap(self, height: List[int]) -> int:
        # 2026-04-12
        # Time 02:41

        pre = suf = ans = 0
        l, r = 0, len(height) - 1

        while l < r:
            pre = max(pre, height[l])
            suf = max(suf, height[r])

            if pre < suf:
                ans += pre - height[l]
                l += 1
            else:
                ans += suf - height[r]
                r -= 1
        return ans

# 前后缀分解，前后缀都是最大值，所以可以使用一个整数记录。
# pre 和 suf包含自己，这样不用考虑引入负数的判断，并且相遇点不用计算（最高处），就不用写l<=r了。
```

```python
# 94. 二叉树的中序遍历
# E
# 5

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def inorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        # 2026-04-12
        # Time 02:11
        # 递归实现
        ans = []
        def inorder(node):
            if not node:
                return None

            inorder(node.left)
            ans.append(node.val)
            inorder(node.right)
        inorder(root)
        return ans
```
