```python
# 189. 轮转数组
# M
# 1 又忘记写取模（第三种做法）

class Solution:
    def rotate(self, nums: List[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        # 2026-04-14
        # Time 1:55
        
        def reverse(l, r):
            while l < r:
                nums[l], nums[r] = nums[r], nums[l]
                l += 1
                r -= 1
        n = len(nums)
        k %= n

        reverse(0, n - 1)
        reverse(0, k - 1)
        reverse(k, n - 1)
```

```python
# 438. 找到字符串中所有字母异位词
# M
# 5

class Solution:
    def findAnagrams(self, s: str, p: str) -> List[int]:
        # 2026-04-14
        # Time 06:23

        cnt = Counter(p)  # Counter要比defaultdict慢
        ans = []
        left = 0
        for right, c in enumerate(s):
            cnt[c] -= 1
            while cnt[c] < 0:
                cnt[s[left]] += 1
                left += 1
            if right - left + 1 == len(p):  # 窗口里面没有多出的元素，且长度和p一致，一定满足条件。
                ans.append(left)
        return ans
        
```

```python
# 15. 三数之和
# M
# 5

class Solution:
    def threeSum(self, nums: list[int]) -> list[list[int]]:
        # 2026-04-14
        # Time 4:53
       
        nums.sort()
        n = len(nums)
        ans = []
        for i in range(n - 2):
            if i > 0 and nums[i] == nums[i-1]:
                continue
            if nums[i] + nums[n-1] + nums[n-2] < 0:
                continue
            if nums[i] + nums[i+1] + nums[i+2] > 0:
                break

            l, r = i + 1, n - 1
            while l < r:
                s = nums[i] + nums[l] + nums[r]
                if s > 0:
                    r -= 1
                elif s < 0:
                    l += 1
                else:
                    ans.append([nums[i], nums[l], nums[r]])
                    x = l + 1
                    while x < r and nums[x] == nums[l]:
                        x += 1
                    l = x
                    y = r - 1
                    while l < y and nums[r] == nums[y]:
                        y -= 1
                    r = y
        return ans

# 重点是去重那块
```

```python
# 118. 杨辉三角
# E
# 5

class Solution:
    def generate(self, numRows: int) -> List[List[int]]:
        # 2026-04-14
        # Time 3:01

        ans = [[1] * i for i in range(1, numRows + 1)] 
        for i in range(numRows):
            for j in range(1, i):
                ans[i][j] = ans[i-1][j-1] + ans[i-1][j]
        return ans
```

```python
# 198. 打家劫舍
# M
# 5 2 (空间复杂度O(1)做法)

class Solution:
    def rob(self, nums: List[int]) -> int:
        # 2026-04-14
        # Time 6:14 
        # 空间复杂度O(n)

        n = len(nums)
        if n <= 2:
            return max(nums)
        dp = [0] * n
        dp[0] = nums[0]
        dp[1] = max(nums[0], nums[1])

        for i in range(2, n):
            dp[i] = max(dp[i-2] + nums[i], dp[i-1])
        return dp[-1]


class Solution:
    def rob(self, nums: List[int]) -> int:
        # 2026-04-14
        # Time 6:14
        # 空间复杂度O(1)
        
        f1 = f0 = 0
        for i, x in enumerate(nums):
            new_f = max(f1, f0 + x)
            f0 = f1
            f1 = new_f
        
        return f1
```

```python
# 136. 只出现一次的数字
# E
# 5

class Solution:
    def singleNumber(self, nums: List[int]) -> int:
        # 2026-04-14
        # Time 1:00
        ans = 0
        for i in nums:
            ans ^= i
        return ans
```

```python
# 54. 螺旋矩阵
# M
# 2

class Solution:
    def spiralOrder(self, matrix: List[List[int]]) -> List[int]:
        # 2026-04-14
        DIRS = (0, 1), (1, 0), (0, -1), (-1, 0)

        m, n = len(matrix), len(matrix[0])

        i, j, di = 0, -1, 0
        ans = []
        size = m * n
        while len(ans) < size:
            for _ in range(n):
                i += DIRS[di][0]
                j += DIRS[di][1]
                ans.append(matrix[i][j])
            di = (di + 1) % 4  # 下一个方向
            n, m = m - 1, n  # 两两交替 n -> m - 1 -> n - 1 -> m - 2
        return ans

# 主要还是要找到规律。
```

```python
# 48. 旋转图像
# M
# 1

class Solution:
    def rotate(self, matrix: List[List[int]]) -> None:
        """
        Do not return anything, modify matrix in-place instead.
        """
        # 2026-04-14
        # 顺时针旋转90度->第j列元素去了第j行，第i行的元素去到第n-1-i列。
        # 旋转可以通过先转置，再行翻转实现。

        n = len(matrix)
        # 1. 转置
        for i in range(n):
            for j in range(i):
                matrix[i][j], matrix[j][i] = matrix[j][i], matrix[i][j]
        # 2. 翻转
        for row in matrix:
            row.reverse()
        
```

```python
# 141. 环形链表
# E
# 5

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, x):
#         self.val = x
#         self.next = None

class Solution:
    def hasCycle(self, head: Optional[ListNode]) -> bool:
        fast = slow = head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            if fast is slow:
                return True
        return False
```

```python
# 234. 回文链表
# E
# 1

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def isPalindrome(self, head: Optional[ListNode]) -> bool:
        # Time 16:26
        slow = fast = head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
        
        mid = slow

        pre, cur = None, head
        cur = mid
        while cur:
            nxt = cur.next
            cur.next = pre
            pre = cur
            cur = nxt

        head2 = pre

        while head2:  # 既然长度不一致，判断短的
            if head.val != head2.val:
                return False
            head = head.next
            head2 = head2.next
        return True


# 这题思路知道，但是卡在了最后两个序列长度不一致上。
# 其实只要判断短的走完就行。
```
