```python
# 54. 螺旋矩阵
# M
# 3
# 上次备注: 难点：边界控制和规律寻找

class Solution:
    def spiralOrder(self, matrix: List[List[int]]) -> List[int]:
        # 2026-04-14
        # Time 11:52
        DIRS = (0, 1), (1, 0), (0, -1), (-1, 0)

        m, n = len(matrix), len(matrix[0])
        di = 0
        i, j = 0, -1
        ans = []
        while n:
            dx, dy = DIRS[di]
            for _ in range(n):
                i += dx
                j += dy
                ans.append(matrix[i][j])

            n, m = m - 1, n
            di = (di + 1) % 4
        return ans

# n, m = m - 1, n 的理解：m只是为了记录n，n下一步要变成m - 1。
```

```python
# 234. 回文链表
# E
# 4
# 上次备注: 卡在最后两个序列长度不一致的处理上

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def isPalindrome(self, head: Optional[ListNode]) -> bool:
        # 2026-04-15
        # Time 07:07

        slow = fast = head
        while fast and fast.next:
            fast = fast.next.next
            slow = slow.next

        pre = None
        cur = slow
        while cur:
            nxt = cur.next
            cur.next = pre
            pre = cur
            cur = nxt

        l1 = head
        l2 = pre
        while l2:  # l2长度大于等于l1
            if l1.val != l2.val:
                return False
            l1 = l1.next
            l2 = l2.next
        return True
```

```python
# 2. 两数相加
# M
# 3

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode], carry=0) -> Optional[ListNode]:
        # 2026-04-15
        # Time 35:09
        cur = dummy = ListNode()
        carry = 0
        while l1 and l2 or carry:
            if l1:
                carry += l1.val
                l1 = l1.next
            if l2:
                carry += l2.val
                l2 = l2.next
            cur.next = ListNode(carry % 10)
            carry //= 10
            cur = cur.next
        return dummy.next
        

```

```python
# 19. 删除链表的倒数第 N 个结点
# M
# 5 

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def removeNthFromEnd(self, head: Optional[ListNode], n: int) -> Optional[ListNode]:
        # 2026-04-15
        # Time 05:47
        # 可能影响头节点的，可以使用dummy简化代码
        
        # 左端点在链表头部，右端点在正数第n个节点。向右移动尺子，当尺子右端点到达链表末尾时，左端点就在倒数第n个节点。
        left = right = dummy = ListNode(next=head)
        for _ in range(n):
            right = right.next

        while right.next:
            left = left.next
            right = right.next
        left.next = left.next.next
        
        return dummy.next
```

```python
# 24. 两两交换链表中的节点
# M
# 2

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def swapPairs(self, head: Optional[ListNode]) -> Optional[ListNode]:
        # 2026-04-15

        node0 = dummy = ListNode(next=head)
        node1 = head
        # 卡在最后越界的细节上面        
        # 必须后面有两个节点
        while node1 and node1.next:
            node2 = node1.next
            node3 = node2.next  # 最远只会用到节点2.next 所以最多只是None 不会报错

            node0.next = node2
            node2.next = node1
            node1.next = node3

            node0 = node1
            node1 = node3
        return dummy.next
```

```python
# 92. 反转链表 II
# M
# 1

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reverseBetween(self, head: Optional[ListNode], left: int, right: int) -> Optional[ListNode]:
        # 2026-04-15
        # Time 10:11
        dummy = p0 = ListNode(next=head)
        
        for _ in range(left - 1):
            p0 = p0.next
        
        pre = None
        cur = p0.next
        for _ in range(right - left + 1):
            nxt = cur.next
            cur.next = pre
            pre = cur
            cur = nxt
        
        # ！！！
        p0.next.next = cur
        p0.next = pre
        return dummy.next
```

```python
# 25. K 个一组翻转链表
# H
# 1


# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reverseKGroup(self, head: Optional[ListNode], k: int) -> Optional[ListNode]:
        # 2026-04-15
        # Time 08:06
        dummy = p0 = ListNode(next=head)
        cur = head
        cnt = 0
        while cur:
            cur = cur.next
            cnt += 1
        
        pre = None
        cur = head
        while cnt >= k:
            cnt -= k
            for _ in range(k):
                nxt = cur.next
                cur.next = pre
                pre = cur
                cur = nxt
            
            nxt = p0.next
            p0.next.next = cur
            p0.next = pre
            p0 = nxt
        
        return dummy.next

# pre被赋的第一个值用不到，所以随便赋值。后面虽然会用到pre，但此时pre已经被cur更新过了（上一个cur）所以不会有问题.
```

```python
# 139. 单词拆分
# M
# 4 (虽然写出来了，但是方法不对，应该枚举长度)

class Solution:
    def wordBreak(self, s: str, wordDict: List[str]) -> bool:
        # 2026-04-15
        # Time: 11:51
        max_len = max(map(len, wordDict))
        words = set(wordDict)

        n = len(s)
        f = [True] + [False] * n
        for i in range(1, n + 1):
            for j in range(i - 1, max(i - max_len - 1, -1), -1):
                if f[j] and s[j:i] in words:
                    f[i] = True
                    break
        
        return f[n]
# 本题 wordDict 至多有 1000 个字符串，但最多只有 20 种不同的长度，所以应该枚举长度，而不是枚举 wordDict 中的字符串。
```
