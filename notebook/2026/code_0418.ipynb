```python
# 24. 两两交换链表中的节点
# M
# 5
# 上次备注: 卡在越界细节，须保证后面有两个节点


# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def swapPairs(self, head: Optional[ListNode]) -> Optional[ListNode]:
        # 2026-04-18
        # Time 05:12

        dummy = p0 = ListNode(next=head)
        pre = None
        cur = head
        while cur and cur.next:
            for _ in range(2):
                nxt = cur.next
                cur.next = pre
                pre = cur
                cur = nxt
            
            nxt = p0.next
            p0.next.next = cur
            p0.next = pre
            p0 = nxt
        return dummy.next

```

```python
# 300. 最长递增子序列
# M
# 5
# 上次备注: 需熟练掌握 nlogn 的贪心+二分法

class Solution:
    def lengthOfLIS(self, nums: List[int]) -> int:
        # 2026-04-18
        # Time 04:26
        # 贪心 + 二分
        g = []
        for i, x in enumerate(nums):
            j = bisect_left(g, x)
            if j == len(g):
                g.append(x)
            else:
                g[j] = x
        return len(g) 
```

```python
# 206. 反转链表
# E
# 4

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        # 2026-04-18
        # Time 3:45
        pre = None
        cur = head
        while cur:
            nxt = cur.next
            cur.next = pre
            pre = cur
            cur = nxt

        return pre


# 这题一开始还习惯性地写了个dummy节点，但是由于整个链表翻转，所以返回pre即可（开头是确定的，不需要靠dummy）
# 如果加了dummy，那么翻转后，指向的是最后一个节点。
```

```python
# 41. 缺失的第一个正数
# H
# 3

class Solution:
    def firstMissingPositive(self, nums: List[int]) -> int:
        # 2026-04-18
        # Time 19:41

        n = len(nums)
        for i in range(n):
            while 1 <= nums[i] <= n and nums[nums[i] - 1] != nums[i]:
                j = nums[i] - 1
                nums[i], nums[j] = nums[j], nums[i]

        for i in range(n):
            if nums[i] != i + 1:
                return i + 1
        return n + 1

# 遗忘了多次出现相同数的判断；并且把判断放在了交换后导致数组越界。
```

```python
# 560. 和为 K 的子数组
# M
# 5

class Solution:
    def subarraySum(self, nums: List[int], k: int) -> int:
        # 2026-04-18
        # Time 02:11

        cnt = defaultdict(int)
        cnt[0] = 1
        pre = 0
        ans = 0
        for x in nums:
            pre += x
            ans += cnt[pre-k]
            cnt[pre] += 1
        return ans
```

```python
# 239. 滑动窗口最大值
# H
# 3

class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        # 2026-04-18
        # Time 08:39

        q = deque()
        ans = []
        for i, x in enumerate(nums):
            while q and nums[q[-1]] <= x:
                q.pop()
            
            q.append(i)

            l = i - k + 1
            if l >= 0:
                ans.append(nums[q[0]])

            if q[0] <= l:
                q.popleft()

        return ans
       


# 两个条件：1. left大于0，更新答案 2. q[0]小于等于left就直接弹出
```
