```python
# 25. K 个一组翻转链表
# H
# 5

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reverseKGroup(self, head: Optional[ListNode], k: int) -> Optional[ListNode]:
        # 2026-04-19
        # Time 3:41
        
        dummy = p0 = ListNode(next=head)
        cur = head
        cnt = 0
        while cur:
            cnt += 1
            cur = cur.next

        pre = None
        cur = head
        while cnt >= k:
            cnt -= k
            for _ in range(k):
                nxt = cur.next
                cur.next = pre
                pre = cur
                cur = nxt

            nxt = p0.next
            p0.next.next = cur
            p0.next = pre
            p0 = nxt

        return dummy.next
```

```python
# 54. 螺旋矩阵
# M
# 4
# 上次备注: 注意 n, m = m - 1, n 的交替规律


class Solution:
    def spiralOrder(self, matrix: List[List[int]]) -> List[int]:
        # 2026-04-19
        # Time 06:08
        DIRS = (0, 1), (1, 0), (0, -1), (-1, 0)

        m, n = len(matrix), len(matrix[0])

        di = 0
        i, j = 0, -1
        ans = []
        while n:
            dx, dy = DIRS[di]
            for _ in range(n):
                i += dx
                j += dy
                ans.append(matrix[i][j])
            
            di = (di + 1) % 4
            n, m = m - 1, n
        return ans



# 还是不能直观的去理解n, m = m - 1, n
# 分两步理解：1.n, m = m, n 方向转了90°，横变竖，竖变横
# 2. 把那个正要变成“当前步数”的m减去 1，即 m - 1。
```

```python
# 48. 旋转图像
# M
# 5
# 上次备注: 先转置再逆序


class Solution:
    def rotate(self, matrix: List[List[int]]) -> None:
        """
        Do not return anything, modify matrix in-place instead.
        """
        # 2026-04-19
        # Time 02:42

        # i, j ->  n - j - 1, i
        n = len(matrix)
        for i in range(n):
            for j in range(i):
                matrix[i][j], matrix[j][i] = matrix[j][i], matrix[i][j]
        
        for row in matrix:
            row.reverse()

# i, j ->  n - j - 1, i
```

```python
# 17. 电话号码的字母组合
# M
# 5

class Solution:
    def letterCombinations(self, digits: str) -> List[str]:
        # 2026-04-19
        # Time 6:43

        # 不用恢复现场
        mapping = ['','', 'abc', 'def', 'ghi', 'jkl', 'mno', 'pqrs', 'tuv', 'wxyz']
        n = len(digits)
        path = [0] * n
        ans = []
        def dfs(i):
            if i == n:
                ans.append("".join(path))
                return
            
            for k in mapping[int(digits[i])]:
                path[i] = k
                dfs(i + 1)
            i += 1 
            
        dfs(0)
        return ans
```

```python
# 131. 分割回文串
# M
# 3 4

class Solution:
    def partition(self, s: str) -> List[List[str]]:
        # 2026-04-19
        # 选或不选
        n = len(s)
        ans = []
        path = []


        def dfs(i: int, start: int) -> None:
            if i == n:
                ans.append(path.copy())
                return
            
            # 不选
            if i < n - 1:  # i=n-1时必须分割
                dfs(i + 1, start)
            
            # 分割 得到子串 [start, i]
            t = s[start: i + 1]
            if t == t[::-1]:
                path.append(t)
                dfs(i + 1, i + 1)
                path.pop()

        dfs(0, 0)
        return ans


class Solution:
    def partition(self, s: str) -> List[List[str]]:
        # 2026-04-19
        # Time 09:32
        # 枚举选哪个

        ans = []
        path = []
        n = len(s)

        def dfs(i):
            if i == n:
                ans.append(path.copy())
                return

            for j in range(i, n):
                t = s[i:j + 1]
                if t == t[::-1]:
                    path.append(t)
                    dfs(j + 1)
                    path.pop()
            
        dfs(0)
        return ans
```
