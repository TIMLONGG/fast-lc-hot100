```python
# 240. 搜索二维矩阵 II
# M
# 5
# 上次备注: 右上搜索到左下，排除法

class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:
        # 2026-04-20
        # Time 2:44


        m, n = len(matrix), len(matrix[0])
        i, j = 0, n - 1
        while i < m and j >= 0:
            x = matrix[i][j]
            if x == target:
                return True
            if x > target:
                j -= 1
            else:
                i += 1
        return False
```

```python
# 234. 回文链表
# E
# 4
# 上次备注: 翻转后半部分，判断短的走完即可


# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def isPalindrome(self, head: Optional[ListNode]) -> bool:
        # 2026-04-20
        # Time 10:06

        fast = slow = head
        while fast and fast.next:
            fast = fast.next.next
            slow = slow.next
        
        cur = slow
        pre = None
        while cur:
            nxt = cur.next
            cur.next = pre
            pre = cur
            cur = nxt
        
        node2 = pre
        while node2:
            if head.val != node2.val:
                return False
            head = head.next
            node2 = node2.next
        return True
        
```

```python
# 92. 反转链表 II
# M
# 4

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reverseBetween(self, head: Optional[ListNode], left: int, right: int) -> Optional[ListNode]:
        # 2026-04-20
        # Time 07:52
        
        dummy = p0 = ListNode(next=head)
        
        cur = head
        for _ in range(left - 1):
            p0 = p0.next
        pre = None
        cur = p0.next
        for _ in range(right - left + 1):
            nxt = cur.next
            cur.next = pre
            pre = cur
            cur = nxt
        p0.next.next = cur
        p0.next = pre
        return dummy.next
```

```python
# 543. 二叉树的直径
# E
# 4

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def diameterOfBinaryTree(self, root: Optional[TreeNode]) -> int:
        # 2026-04-20
        # Time 12:34
        ans = 0
        def dfs(node):
            if node is None:
                return -1
            nonlocal ans
            l_left = dfs(node.left)
            l_right = dfs(node.right)
            ans = max(ans, l_left + l_right + 2)

            return max(l_left, l_right) + 1
            
        dfs(root)
        return ans
        
# 确实想了有一会，不过最后还是做出来了。核心是区分链长和直径。一边计算经过这个节点的直径，一边返回这个节点到叶子节点的最大链长。
```

```python
# 108. 将有序数组转换为二叉搜索树
# E
# 5

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def sortedArrayToBST(self, nums: List[int]) -> Optional[TreeNode]:
        # 2026-04-20
        # Time 4:08

        def dfs(l, r):
            if l > r:
                return
            mid = (l + r) // 2
            root = TreeNode(nums[mid])
            root.left = dfs(l, mid - 1)
            root.right = dfs(mid + 1, r)
            return root
        
        return dfs(0, len(nums) - 1)
```

```python
# 39. 组合总和
# M
# 4

class Solution:
    def combinationSum(self, candidates: List[int], target: int) -> List[List[int]]:
        # 2026-04-20
        # Time 05:10
        candidates.sort()
        # 选或不选
        ans = []
        path = []
        n = len(candidates)
        def dfs(i, s):            
            if s == target:
                ans.append(path.copy())
                return
            
            if i == n or target - s < candidates[i]:
                return
            
            # 不选
            dfs(i + 1, s)

            # 选
            path.append(candidates[i])
            dfs(i, s + candidates[i])
            path.pop()
        
        
        dfs(0, 0)
        return ans
# 剪枝非常重要
```

```python
# 22. 括号生成
# M
# 3
class Solution:
    def generateParenthesis(self, n: int) -> List[str]:
        # 2026-04-20
        # Time 22:49
 
        m = n * 2
        ans = []
        path = [0] * m
        
        def dfs(left, right):
            if right == n:
                ans.append(''.join(path))
                return

            if left < n:
                path[left + right] = '('
                dfs(left + 1, right)
            
            if right < left:
                path[left + right] = ')'
                dfs(left, right + 1)
        
        dfs(0, 0)
        return ans
# 卡在了右括号何时能选的地方
```

```python
# 46. 全排列
# M
# 4

class Solution:
    def permute(self, nums: List[int]) -> List[List[int]]:
        # 2026-04-20
        # Time 04:24

        ans = []
        path = []
        st = set()
        n = len(nums)
        def dfs(i):
            if i == n:
                ans.append(path.copy())
                return
            
            for j in nums:
                if j not in st:
                    path.append(j)
                    st.add(j)
                    dfs(i + 1)
                    st.remove(j)
                    path.pop()
        dfs(0)
        return ans
```

```python
# 51. N 皇后
# H
# 2


class Solution:
    def solveNQueens(self, n: int) -> List[List[str]]:
        # 2026-04-20
        # Time 15:14

        ans = []
        col = [0] * n
        main_diag = set()
        anti_diag = set()

        def isValid(r, c):
            if r - c in main_diag or r + c in anti_diag:
                return False
            return True
        
        def dfs(i, s):
            if i == n:
                ans.append(['.' * c + 'Q' + '.' * (n-c-1) for c in col])
                return
            
            for j in s:
                if isValid(i, j):
                    col[i] = j
                    main_diag.add(i - j)
                    anti_diag.add(i + j)
                    dfs(i + 1, s - {j})
                    main_diag.remove(i - j)
                    anti_diag.remove(i + j)
        
        dfs(0, set(range(n)))
        return ans


# main和anti_diag的优化，不要忘记恢复现场。
```

```python
# 79. 单词搜索
# M
# 1


class Solution:
    def exist(self, board: List[List[str]], word: str) -> bool:
        cnt = Counter(c for row in board for c in row)
        if not cnt >= Counter(word):
            return False
        if cnt[word[-1]] < cnt[word[0]]:
            word = word[::-1]
        
        m, n = len(board), len(board[0])
        
        def dfs(i, j, s):
            if board[i][j] != word[s]:
                return False
            if s == len(word) - 1:
                return True

            board[i][j] = ''
            for di, dj in [(0, 1), (1, 0), (0, -1), (-1, 0)]:
                ni, nj = i + di, j + dj
                if 0 <= ni < m and 0 <= nj < n and dfs(ni, nj, s + 1):
                    return True
            board[i][j] = word[s]
            return False
        return any(dfs(i, j, 0) for i in range(m) for j in range(n))

# 两个优化很关键，需要记住
```
