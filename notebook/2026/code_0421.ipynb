```python
# 51. N 皇后
# H
# 2
# 上次备注: main和anti_diag的优化，不要忘记恢复现场

class Solution:
    def solveNQueens(self, n: int) -> List[List[str]]:
        # 2026-04-21
        # Time 21:11

        diag1 = [False] * (n * 2 - 1) 
        diag2 = [False] * (n * 2 - 1)
        ans = []
        visited = [False] * n
        col = [0] * n

        def dfs(i):
            if i == n:
                ans.append(['.' * c + 'Q' + '.' * (n-c-1) for c in col])
                return

            for j in range(n):
                if not visited[j] and not diag1[i + j] and not diag2[i - j]:
                    col[i] = j
                    visited[j] = diag1[i + j] = diag2[i - j] = True
                    dfs(i + 1)
                    visited[j] = diag1[i + j] = diag2[i - j] = False
        dfs(0)
        return ans

# 依然不够熟练
```
