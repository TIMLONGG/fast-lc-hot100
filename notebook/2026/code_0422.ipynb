```python
# 79. 单词搜索
# M
# 3
# 上次备注: 两个优化：Counter预判字符数量、翻转稀有端到前


class Solution:
    def exist(self, board: List[List[str]], word: str) -> bool:
        # 2026-04-22 
        # Time 21:40

        # 优化1 
        cnt = Counter(c for row in board for c in row)
        if not cnt >= Counter(word):
            return False
        
        # 优化2
        if cnt[word[0]] <= cnt[word[-1]]:
            word = word[::-1]
        
        m, n = len(board), len(board[0])
        def dfs(i, j, k):
            x = board[i][j]
            if x != word[k]:
                return False
            
            if k == len(word) - 1:
                return True

            board[i][j] = ''
            for ni, nj in [(i, j + 1), (i + 1, j), (i, j - 1), (i - 1, j)]:
                if 0 <= ni < m and 0 <= nj < n and dfs(ni, nj, k + 1):  # 将子任务的结果判断放在这里
                    return True
            board[i][j] = x

            return False

        return any(dfs(i, j, 0) for i in range(m) for j in range(n))


# 当时在卡在了循环里面return的逻辑，写成了return dfs(ni, nj, k + 1)，但是这样只会一直遍历第一个方向。
# 所以需要判断 + return True 
```

```python
# 322. 零钱兑换
# M
# 5

class Solution:
    def coinChange(self, coins: List[int], amount: int) -> int:
        # 2026-04-22
        # Time 4:23

        dp = [inf] * (amount + 1)
        dp[0] = 0
        for i in range(1, amount + 1):
            for j in coins:
                if i < j:
                    continue
                dp[i] = min(dp[i], dp[i - j] + 1)
        ans = dp[-1]
        return ans if ans < inf else -1
```

```python
# 2. 两数相加
# M
# 2 2

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
        # 2026-04-22
        # Time 

        cur = dummy = ListNode()
        c = 0
        while l1 or l2 or c:
            if l1:
                c += l1.val
                l1 = l1.next
            if l2:
                c += l2.val
                l2 = l2.next
            cur.next = ListNode(c % 10)
            c //= 10
            cur = cur.next
        return dummy.next

# 一直想用循环进行原地修改的，但是感觉非常麻烦，遂放弃。
# 这里用了dummy节点，然后每一个迭代新建一个节点，我感觉这样不够优雅。

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode], c:int=0) -> Optional[ListNode]:
        # 2026-04-22
        # Time 
        
        if l1 is None and l2 is None:
            return ListNode(c) if c else None
        
        if l1 is None:
            l1, l2 = l2, l1
        
        s = c + l1.val + (l2.val if l2 else 0)
        l1.val = s % 10
        l1.next = self.addTwoNumbers(l1.next, l2.next if l2 else None, s // 10)
        return l1

# 递归方法需要好好掌握
```
