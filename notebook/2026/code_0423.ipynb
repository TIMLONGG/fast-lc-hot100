```python
# 51. N 皇后
# H
# 3
# 上次备注: 依然不够熟练

class Solution:
    def solveNQueens(self, n: int) -> List[List[str]]:
        # 2026-04-23
        # Time 14:15

        m = 2 * n - 1
        main_diag = [False] * m
        anti_diag = [False] * m
        visited = [False] * n
        col = [0] * n
        ans = []
        def dfs(i):
            if i == n:
                ans.append(['.' * c + 'Q' + '.' * (n-1-c) for c in col])
                return
            
            for j in range(n):
                if not visited[j] and not main_diag[i - j] and not anti_diag[i + j]:
                    visited[j] = main_diag[i - j] = anti_diag[i + j] = True
                    col[i] = j
                    dfs(i + 1)
                    visited[j] = main_diag[i - j] = anti_diag[i + j] = False

        dfs(0)
        return ans

# 索引那里，python可以使用负数索引，其他语言不行
```

```python
# 189. 轮转数组
# M
# 4

class Solution:
    def rotate(self, nums: List[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        # 2026-04-23
        # Time 1:42

       
        def reverse(l, r):
            while l < r:
                nums[l], nums[r] = nums[r], nums[l]
                l += 1
                r -= 1
        
        n = len(nums)
        k %= n
        reverse(0, n - 1)
        reverse(0, k - 1)
        reverse(k, n - 1)

```

```python
# 21. 合并两个有序链表
# E
# 5
# 上次备注: 虚拟头节点简化原地操作难度

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def mergeTwoLists(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
        # 2026-04-23
        # Time 05:35
        dummy = p0 = ListNode()

        while list1 and list2:
            if list1.val <= list2.val:
                p0.next = list1
                list1 = list1.next
            else:
                p0.next = list2
                list2 = list2.next
            p0 = p0.next
        p0.next = list1 or list2
        return dummy.next
            
```

```python
# 56. 合并区间
# M
# 4

class Solution:
    def merge(self, intervals: List[List[int]]) -> List[List[int]]:
        # 2026-04-23
        # Time 04:03

        intervals.sort(key=lambda x: x[0])
        ans = []
        for x in intervals:
            l, r = x[0], x[1]
            if ans and l <= ans[-1][1]:
                ans[-1][1] = max(ans[-1][1], r)
            else:
                ans.append([l, r])
        return ans

# 排序的lambda表达式： lambda x: x[0]
```

```python
# 79. 单词搜索
# M
# 4
# 上次备注: 卡点：循环里不能直接 `return dfs(...)`，否则只会走第一个方向；应判断后再 `return True`


class Solution:
    def exist(self, board: List[List[str]], word: str) -> bool:
        # 2026-04-23
        # Time 04:56

        cnt = Counter(c for row in board for c in row)
        if not cnt >= Counter(word):
            return False
        if cnt[word[0]] < cnt[word[-1]]:
            word = word[::-1]
        m, n = len(board), len(board[0])
        def dfs(i, j, k):
            x = board[i][j]
            if x != word[k]:
                return False

            if k == len(word) - 1:
                return True
            board[i][j] = ''
            for ni, nj in (i, j + 1), (i + 1, j), (i, j - 1), (i - 1, j):
                if 0 <= ni < m and 0 <= nj < n and dfs(ni, nj, k + 1):
                    return True
            board[i][j] = x
            return False
        
        return any(dfs(i, j, 0) for i in range(m) for j in range(n))
```
