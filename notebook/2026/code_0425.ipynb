```python
# 131. 分割回文串
# M
# 3

class Solution:
    def partition(self, s: str) -> List[List[str]]:
        # 2026-04-25
        # Time 16:14
        
        ans = []
        path = []
        n = len(s)

        # 现在 s 未被分割的部分为 [k, n-1]
        # 当前位于下标 i，讨论是否在 i 和 i+1 之间切一刀
        def dfs(i, k):
            if i == n:
                ans.append(path.copy())
                return

            
            if i < n - 1:
                dfs(i + 1, k)

            x = s[k:i+1]
            if x == x[::-1]:
                path.append(x)
                dfs(i + 1, i + 1)
                path.pop()
        dfs(0, 0)
        return ans

# path.append(x) 而不是 path.append("".join(x)) -> O(n)
```

```python
# 77. 组合
# M
# 4 4
# 上次备注: 倒序代码更简单


class Solution:
    def combine(self, n: int, k: int) -> List[List[int]]:
        # 2026-04-25
        # Time 05:26

        path = []
        ans = []

        # 选或不选
        def dfs(i):
            s = k - len(path)
            if s == 0:
                ans.append(path.copy())
                return
            
            if i > s:
                dfs(i - 1)
            
            path.append(i)
            dfs(i - 1)
            path.pop()
        
        dfs(n)
        return ans


class Solution:
    def combine(self, n: int, k: int) -> List[List[int]]:
        # 2026-04-25
        # Time 04:27

        path = []
        ans = []

        # 枚举选哪个
        def dfs(i):
            s = k - len(path)  # 还要选s个数
            if s == 0:
                ans.append(path.copy())
                return
            
            # 下一个数必须大于等于s，否则后面不够选
            for j in range(i, s - 1, -1):
                path.append(j)
                dfs(j - 1)
                path.pop()
        
        dfs(n)
        return ans

# 两种做法都需要掌握，尤其是枚举选哪个的剪枝
```

```python
# 22. 括号生成
# M
# 4
# 上次备注: 卡在了右括号何时能选的地方


class Solution:
    def generateParenthesis(self, n: int) -> List[str]:
        # 2026-04-25
        # Time 4:42
 
        ans = []
        m = 2 * n
        path = [0] * m

        def dfs(l, r):
            if r == n:
                ans.append("".join(path))
                return
            
            if l < n:
                path[l + r] = '('
                dfs(l + 1, r)
            
            if r < l:
                path[l + r] = ')'
                dfs(l, r + 1)
        
        dfs(0, 0)
        return ans

# dfs里面用两个参数即可
```

```python
# 24. 两两交换链表中的节点
# M
# 4

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def swapPairs(self, head: Optional[ListNode]) -> Optional[ListNode]:
        # 2026-04-25
        # Time 06:36

        dummy = p0 = ListNode(next=head)
        n1 = head

        while n1 and n1.next:
            n2 = n1.next
            n3 = n2.next

            p0.next = n2
            n2.next = n1
            n1.next = n3

            p0 = n1
            n1 = n3
        return dummy.next

# 两两交换没必要用反转链表那一套
```
