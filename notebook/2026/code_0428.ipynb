```python
# 239. 滑动窗口最大值
# H
# 3
# 上次备注: 注意先更新答案，再按左边界弹出队首


class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        # 2026-04-28
        # Time 8:37

        q = deque()
        ans = []

        for i, x in enumerate(nums):
            while q and x >= nums[q[-1]]:
                q.pop()
            
            q.append(i)

            left = i - k + 1

            if q[0] < left:
                q.popleft()
            
            if left >= 0:
                ans.append(nums[q[0]])

        return ans

# left 应该通过i来计算，我一开始写成q[-1]了，可能会影响速度
```

```python
# 54. 螺旋矩阵
# M
# 4
# 上次备注: 两步理解 n,m=m-1,n：先交换方向，再将变成"当前步数"的那个减1

class Solution:
    def spiralOrder(self, matrix: List[List[int]]) -> List[int]:
        # 2026-04-19
        # Time 7:23
        DIRS = (0, 1), (1, 0), (0, -1), (-1, 0)

        m, n = len(matrix), len(matrix[0])
        i, j = 0, -1
        di = 0
        ans = []
        while n:
            dx, dy = DIRS[di]
            for _ in range(n):
                i += dx 
                j += dy
                ans.append(matrix[i][j])
                
            di = (di + 1) % 4
            n, m = m -1, n
        return ans
            
```

```python
# 206. 反转链表
# E
# 4
# 上次备注: 整体翻转直接返回 pre，不需要 dummy

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        # 2026-04-28
        # Time 2:39

        cur = head
        pre = None
        
        while cur:
            nxt = cur.next
            cur.next = pre
            pre = cur
            cur = nxt

        return pre

# 又新建dummy了。
```
