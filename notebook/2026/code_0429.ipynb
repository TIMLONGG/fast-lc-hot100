```python
# 2. 两数相加
# M
# 4
# 上次备注: 递归迭代都需要着重加强

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode], c:int=0) -> Optional[ListNode]:
        # 2026-04-39
        # Time 6:55

        # 迭代
        dummy = cur = ListNode()
        while l1 or l2 or c:  
            s = c
            if l1:
                s += l1.val
                l1 = l1.next
            if l2:
                s += l2.val
                l2 = l2.next
            c = s // 10
            s %= 10
            cur.next = ListNode(s)
            cur = cur.next
        
        return dummy.next
```

```python
# 51. N 皇后
# H
# 5
# 上次备注: 索引点：Python 可用负数索引，其他语言不行


class Solution:
    def solveNQueens(self, n: int) -> List[List[str]]:
        # 2026-04-29
        # Time 4:24

        ans = []
        m = 2 * n - 1
        diag1 = [False] * m
        diag2 = [False] * m
        visit = [False] * n
        col = [0] * n

        def dfs(i):
            if i == n:
                ans.append(['.' * c + 'Q' + '.' * (n-c-1) for c in col])
                return
            
            for j in range(n):
                if not visit[j] and not diag1[i - j] and not diag2[i + j]:
                    visit[j] = diag1[i - j] = diag2[i + j] = True
                    col[i] = j
                    dfs(i + 1)
                    visit[j] = diag1[i - j] = diag2[i + j] = False
            
        dfs(0)
        return ans
```

```python
# 300. 最长递增子序列
# M
# 5

class Solution:
    def lengthOfLIS(self, nums: List[int]) -> int:
        # 2026-04-29
        # Time 02:18

        g = []
        for j, x in enumerate(nums):
            j = bisect_left(g, x)
            if j == len(g):
                g.append(x)
            else:
                g[j] = x
        
        return len(g)
```

```python
# 34. 在排序数组中查找元素的第一个和最后一个位置
# M
# 4

class Solution:
    def searchRange(self, nums: List[int], target: int) -> List[int]:
        # 2026-04-29
        # Time: 6:59

        # 这题不能用nums[mid] <= t来判断
        # 因为那样导致l左侧小于等于t（循环不变量），这样得到的是最后一个的位置。
        def lower_bound(t):
            l, r = 0, len(nums) - 1
            while l <= r:
                mid = (l + r) // 2
                if nums[mid] < t:
                    l = mid + 1  # [mid + 1, r]
                else:
                    r = mid - 1  # [l, mid - 1]
            return l
        # [r, l]
        # 因为l左侧是小于t的，r右侧是大于等于t的，所以只能选l，得到第一个大于等于t的元素
        
        l = lower_bound(target)
        if l == len(nums) or nums[l] != target:
            return [-1, -1]
        else:
            return [l, lower_bound(target+1) - 1]

# 我倾向于使用闭区间的写法 比较对称
```
