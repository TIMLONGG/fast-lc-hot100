```python
# 41. 缺失的第一个正数
# H
# 4
# 上次备注: 易错点：重复数字判断遗漏，且交换后判断会越界


class Solution:
    def firstMissingPositive(self, nums: List[int]) -> int:
        # 2026-05-01
        # Time 6:13
        n = len(nums)
        for i, x in enumerate(nums):
            while 1 <= nums[i] < n and nums[nums[i] - 1] != nums[i]:
                j = nums[i] - 1
                nums[j], nums[i] = nums[i], nums[j]
        
        for i in range(n):
            if nums[i] != i + 1:
                return i + 1
        return n + 1
```

```python
# 25. K 个一组翻转链表
# H
# 3

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reverseKGroup(self, head: Optional[ListNode], k: int) -> Optional[ListNode]:
        # 2026-05-01
        # Time 17：22

        dummy = p0 = ListNode(next=head)
        cnt = 0
        cur = head
        while cur:
            cur = cur.next
            cnt += 1
        
        pre = None
        cur = head
        while cnt >= k:
            cnt -= k

            for _ in range(k):
                nxt = cur.next
                cur.next = pre
                pre = cur
                cur = nxt
            
            nxt = p0.next
            p0.next.next = cur
            p0.next = pre
            p0 = nxt
        return dummy.next

# 有些生疏了，不过依然做出来了，最后p0 = nxt 写成 p0 = p0.next 低级错误
```
