```python
# 4. 寻找两个正序数组的中位数
# H
# 4
# 上次备注: hot100 最难题目，核心是固定左半元素数量后通过 i/j 建立二分循环不变量

class Solution:
    def findMedianSortedArrays(self, nums1: List[int], nums2: List[int]) -> float:
        # 2026-05-02
        # Time 17:28

        if len(nums1) > len(nums2):
            nums1, nums2 = nums2, nums1
        
        m, n = len(nums1), len(nums2)
        l, r = 0, m - 1
        # 循环不变量：
        # nums1[l-1] <= nums2[j+1]
        # nums1[r+1] > nums2[j+1]
        while l <= r:
            i = (l + r) // 2
            j = (m + n + 1) // 2 - i - 2
            if nums1[i] <= nums2[j+1]:
                l = i + 1
            else:
                r = i - 1
        
        i = r
        j = (m + n + 1) // 2 - i - 2

        l1 = nums1[i] if i >= 0 else -inf
        l2 = nums2[j] if j >= 0 else -inf
        r1 = nums1[i + 1] if i + 1 < m else inf
        r2 = nums2[j + 1] if j + 1 < n else inf

        maxl = max(l1, l2)
        minr = min(r1, r2)

        return maxl if (m + n) % 2 else (maxl + minr) / 2



# 为什么要调换顺序？
# 如果 nums1 比 nums2 长很多，i 的范围太大，算出来的 j 就可能小于 -1，导致 j + 1 < 0，
```

```python
# 33. 搜索旋转排序数组
# M
# 4

class Solution:
    def search(self, nums: List[int], target: int) -> int:
        # 2026-05-02
        # Time 9:42

        # 寻找数组最小值
        n = len(nums)
        l, r = 0, n - 2
        while l <= r:
            mid = (l + r) // 2
            if nums[mid] < nums[-1]:
                r = mid - 1
            else:
                l = mid + 1
        
        if target > nums[-1]:
            j = bisect_left(nums, target, 0, l)
            if nums[j] != target:
                return -1
            return j
        else:
            j = bisect_left(nums, target, l, n)
            if nums[j] != target:
                return -1
            return j    
```

```python
# 543. 二叉树的直径
# E
# 4
# 上次备注: 核心是区分链长和直径，DFS 一边更新直径一边返回链长

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    
    def diameterOfBinaryTree(self, root: Optional[TreeNode]) -> int:
        # 2026-05-02
        # Time 11:49
        ans = 0
        def dfs(node):
            if node is None:
                return -1
            nonlocal ans
            l_left = dfs(node.left)
            l_right = dfs(node.right)
            ans = max(l_left + l_right + 2, ans)

            return max(l_left, l_right) + 1
        
        dfs(root)
        return ans
```
