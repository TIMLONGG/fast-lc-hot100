```python
# 148. 排序链表
# M
# 5
# 上次备注: 注意寻找中点后要断开左侧链表与中点连接

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def sortList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        # 2026-05-04
        # Time 8:05

        # 寻找中点
        def middleNode(head):
            slow = fast = head
            p0 = ListNode(next=slow)
            while fast and fast.next:
                p0 = p0.next
                slow = slow.next
                fast = fast.next.next
            p0.next = None
            return slow
        
        # 排序两个有序链表
        def sortlist(l1, l2):
            dummy = p0 = ListNode()
            while l1 and l2:
                if l1.val < l2.val:
                    p0.next = l1
                    l1 = l1.next
                else:
                    p0.next = l2
                    l2 = l2.next
                p0 = p0.next
            p0.next = l1 or l2
            return dummy.next
        
        if not head or not head.next:
            return head
        
        head1 = middleNode(head)
        head = self.sortList(head)
        head1 = self.sortList(head1)
        return sortlist(head, head1)

        
```

```python
# 78. 子集
# M
# 5

class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:
        # 2026-05-04
        # Time 1:58


        # 选或不选
        path = []
        ans = []
        n = len(nums)
        def dfs(i):
            if i == n:
                ans.append(path.copy())
                return
            
            # 不选
            dfs(i+1)

            # 选
            path.append(nums[i])
            dfs(i+1)
            path.pop()

        dfs(0)
        return ans


class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:
        # 2026-05-04
        # Time 1:58

        # 枚举选哪个
        path = []
        ans = []
        n = len(nums)

        def dfs(i):
            ans.append(path.copy())
            
            for j in range(i, n):
                path.append(nums[j])
                dfs(j+1)
                path.pop()
        dfs(0)
        return ans

```
