```python
# 1143. 最长公共子序列
# M
# 5

class Solution:
    def longestCommonSubsequence(self, s: str, t: str) -> int:
        # 2026-05-06
        # Time 1:46

        @cache
        def dfs(i, j):
            if i < 0 or j < 0:
                return 0
            
            if s[i] == t[j]:
                return dfs(i - 1, j - 1) + 1
            
            return max(dfs(i - 1, j), dfs(i, j-1))
            
        return dfs(len(s)-1, len(t)-1)
```

```python
# 2. 两数相加
# M
# 5

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode], c:int=0) -> Optional[ListNode]:
        # 2026-05-06
        # Time 4:24

        dummy = cur = ListNode()

        while l1 or l2 or c:
            s = c
            if l1:
                s += l1.val
                l1 = l1.next
            if l2:
                s += l2.val
                l2 = l2.next
            c = s // 10
            cur.next = ListNode(s % 10)
            cur = cur.next
        
        return dummy.next

        
```

```python
# 189. 轮转数组
# M
# 4

class Solution:
    def rotate(self, nums: List[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        # 2026-05-06
        # Time 2:33

        def reverse(left, right):
            while left < right:
                nums[left], nums[right] = nums[right], nums[left]
                left += 1
                right -= 1

        n = len(nums)
        k %= n
        reverse(0, n-1)
        reverse(0, k-1)
        reverse(k, n-1)
```
