```python
# 74. 搜索二维矩阵
# M
# 5
# 上次备注: 关键在坐标转换，也可以用排除法但二分复杂度一般更低

class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:
        # 2026-05-07
        # Time: 14:16
        m, n = len(matrix), len(matrix[0])
        l, r = 0, m * n -1

        while l <= r:
            mid = (l + r) // 2
            i, j = mid // n, mid % n
            x = matrix[i][j]
            if x == target:
                return True
            if x > target:
                r = mid - 1
            else:
                l = mid + 1
        return False
```

```python
# 39. 组合总和
# M
# 5
# 上次备注: 剪枝非常重要

class Solution:
    def combinationSum(self, candidates: List[int], target: int) -> List[List[int]]:
        # 2026-05-07
        # Time 9:44

        candidates.sort()
        n = len(candidates)
        path = []
        ans = []
        def dfs(i, k):
            if k == 0:
                ans.append(path.copy())
                return

            if i == n:
                return
            if k < candidates[i]:
                return
            
            # 不选 
            dfs(i + 1, k)

            # 选
            path.append(candidates[i])
            dfs(i, k - candidates[i])
            path.pop()

        dfs(0, target)
        return ans
```

```python
# 46. 全排列
# M
# 4

class Solution:
    def permute(self, nums: List[int]) -> List[List[int]]:
        # 2026-05-07
        # Time 5:43
        ans = []
        visit = set()
        nums = set(nums)
        n = len(nums)
        path = [0] * n
        def dfs(i):
            if i == n:
                ans.append(path.copy())
                return
            for j in nums - visit:
                path[i] = j
                visit.add(j)
                dfs(i + 1)
                visit.remove(j)
        
        dfs(0)
        return ans
```

```python
# 215. 数组中的第K个最大元素
# M
# 1

class Solution:
    def findKthLargest(self, nums: List[int], k: int) -> int:
        def partition(left, right):
            i = randint(left, right)
            pivot = nums[i]
            nums[i], nums[left] = nums[left], nums[i]
            i, j = left + 1, right
            while True:
                while i <= j and nums[i] < pivot:
                    i += 1
                
                while i <=j and nums[j] > pivot:
                    j -= 1
                
                if i >= j:
                    break
                
                nums[i], nums[j] = nums[j], nums[i]
                i += 1
                j -= 1
            
            nums[left], nums[j] = nums[j], nums[left]
            return j

        n = len(nums)
        target_index = n - k
        left, right = 0, n - 1
        while True:
            i = partition(left, right)
            if i == target_index:
                return nums[i]
            if i > target_index:
                right = i - 1
            else:
                left = i + 1
```

```python
# 912. 排序数组
# M
# 2

class Solution:
    def sortArray(self, nums: List[int]) -> List[int]:
        # 2026-05-07
        # Time
        def partition(left, right):
            i = randint(left, right)
            pivot = nums[i]
            nums[left], nums[i] = nums[i], nums[left]
            i, j = left + 1, right
            while True:
                while i <= j and nums[i] < pivot:
                    i += 1
                while i <= j and nums[j] > pivot:
                    j -= 1
                
                if i >= j:
                    break
                
                nums[i], nums[j] = nums[j], nums[i]
                i += 1
                j -= 1

            nums[left], nums[j] = nums[j], nums[left]
            return j

        def dfs(left, right):
            ordered = True
            for i in range(left, right):
                if nums[i] > nums[i + 1]:
                    ordered = False
                    break
                
            if ordered:
                return 
            
            i = partition(left, right)
            dfs(left, i - 1)
            dfs(i + 1, right)

        dfs(0, len(nums) - 1)
        return nums
```

```python
# 20. 有效的括号
# E
# 5

class Solution:
    def isValid(self, s: str) -> bool:
        # 2026-05-07
        # Time 2:49
        if len(s) % 2:
            return False

        st = []
        mapping = {'}':'{', ']':'[', ')':'('}
        for i in s:
            if i not in mapping:
                st.append(i)
            elif not st or st.pop() != mapping[i]:
                return False
        return not st
```
